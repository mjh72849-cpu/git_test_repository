```python
m,n=map(int,input().split())
lst=[input().split() for i in range(m)]
C=0
lenth=len(lst[0])
def ok(i,j):
    if lst[i][j]!='#':
        return False
    if i==0 and lst[i][j]=='#':
        return True
    for k in range(i-1):
        if lst[k][j]==1:
            return False
    return True
def ways(i,j,num):
    global C
    if num==0:
        C+=1
        return
    if i==lenth:
        return
    if j==lenth:
        return ways(i+1,0,num)
    for k in range(j,lenth):
        if ok(i,k):
            lst[i][k]=1
            ways(i+1,0,num-1)
            lst[i][k]='#'
            ways(i,k+1,num)
            return       
ways(0,0,n)
print(C)
```
